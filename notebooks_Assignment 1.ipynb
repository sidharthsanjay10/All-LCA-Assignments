```python

student = {'Name': 'ABC', 'Division': 10, 'Rollno': 31, 'Age': 20, 'Grade': 'A+'}
student['City'] = 'Kochi'   # add
del student['Age']          # delete
print("1.", student)

keys = ['name', 'division', 'rollno']
values = ['ABC', '10', 31]
student2 = dict(zip(keys, values))
print("2.", student2)

student3 = {'name': 'ABC', 'panel': 'B', 'rollno': 34, 'marks': [65, 87, 67, 94]}
print("3.", sorted(student3))

student4 = {'name': 'ABC', 'division': '10', 'rollno': 31}
print("4.", student4.keys())
print("4.", student4.values())

marks = {'marks1': 23, 'marks2': 67, 'marks3': 123, 'marks4': 43, 'marks5': 39}
mean = sum(marks.values()) / len(marks)
print("5. Mean =", mean)

students = {'name': ['Yash', 'Neel', 'Dev'], 'rollno': [11, 12, 13], 'marks': [78, 56, 98]}
print("6.", students['name'][2])      # Dev
print("6.", students['rollno'][1])    # 12
print("6.", max(students['marks']))   # 98

word = "hello"
print("7.", end=" ")
for letter in set(word):
    print(f"{letter}:{word.count(letter)}", end="  ")
print()

num, den = 25, 4
print("8.", num // den, num % den)

from datetime import date

date1 = date(2008, 11, 18)
date2 = date(2006, 2, 15)
print("9. Days between dates =", abs((date2 - date1).days))
```
